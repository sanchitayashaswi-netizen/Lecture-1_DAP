```python
# Number Type
# 1. Integer
age = 25
print(age)
print(type(age))

# 2. Float
price = 99.50
print(price)
print(type(price))

# 3. Complex
z = 3 + 4j
print(z)
print(type(z))
```

```python
# Sequence
# 4. String
name = "Amit"
print(name)
print(name[0])
```

```python
# 5. List
marks = [80, 75, 90]
print(marks)
print(marks[1])
```

```python
# 6. Tuple
months = ("Jan", "Feb", "Mar")
print(months)
print(months[0])
```

```python
# 7. Range
numbers = range(1, 6)
for number in numbers:
  print(number)
```

```python
# 8. Arithmetic operators
a = 10
b = 3

print(a=b)
print(a-b)
print(a*b)
print(a/b)
print(a%b)
print(a//b)
print(a**b)
```

```python
# 9. Comparison operators - python
a = 10
b = 5

print(a==b)
print(a!=b)
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
```

```python
# 10. Assignment Operators
x = 10
print(x)     # Assign 10 to x
x += 5
print(x)     # Add 5 to x:x = x + 5
x -= 3
print(x)     # Subtract 3 from x: x = x - 3
x *= 2
print(x)     # Multiply x by 2 x: x = x * 2
x /= 4
print(x)     # Divide x by 4 x: x = x / 4
x %= 3
print(x)
x **= 2
print(x)
```
